# 01 – Erste Schritte

Ein erster Versuch, einen Text "aus der Distanz" zu lesen: Wir zählen Wörter, statt den Text Satz für Satz zu lesen.

Führe die Zellen nacheinander aus (`Shift` + `Enter`).

## 1. Ein Text

Ein kurzer, frei erfundener Beispieltext im Stil einer Theaterkritik.

```python
text = """
Die Bühne ist leer, nur ein Stuhl steht im Licht. Die Inszenierung vertraut
ganz auf das Spiel: Die Schauspielerin trägt den Abend, die Regie hält sich
zurück. Am Ende bleibt der Stuhl im Licht, und das Publikum applaudiert lange.
"""

print(text)
```

## 2. Den Text in Wörter zerlegen

Wir schreiben alles klein, entfernen Satzzeichen und teilen den Text an den Leerzeichen.

```python
import re

woerter = re.findall(r"\w+", text.lower())

print("Anzahl Wörter:", len(woerter))
print("Verschiedene Wörter:", len(set(woerter)))
print(woerter[:10])
```

## 3. Welche Wörter kommen am häufigsten vor?

```python
from collections import Counter

haeufigkeiten = Counter(woerter)

for wort, anzahl in haeufigkeiten.most_common(10):
    print(f"{wort:15} {anzahl}")
```
