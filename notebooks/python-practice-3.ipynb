### 4. From records to a table

pandas could turn `records` into a table directly, but the cells with lists in them (`repTeam`, `articleAuthors`) are hard to work with. Tables work best when every cell holds exactly one value.

So we build our own rows: one dictionary per review, with only the values we want. Where we do not know a value, we leave the key out, and pandas will show an empty cell (`NaN`)

````python
import pandas as pd

rows = []
for record in records:
    row = {
        "article_id": record["article_id"],
        "year": int(record["articleDate"][:4]),
        "play": record["repTitle"],
        "team_size": len(record["repTeam"]),
    }
    if "articleAuthors" in record:
        row["critic"] = record["articleAuthors"][0]
    if "articlePlaces" in record:
        row["place"] = record["articlePlaces"][0]

    # the director: the name of the first pair with the role "Regie"
    for role, name in record["repTeam"]:
        if role == "Regie" and "director" not in row:
            row["director"] = name

    rows.append(row)

reviews = pd.DataFrame(rows)
reviews
````

### 6. A second table: the teams

### Summary

| concept | example |
|---|---|
| a record | `record["repTeam"][0][1]`, `for role, name in record["repTeam"]:` |
| gaps | `if "articleAuthors" in record:`, `len(record["repAuthors"]) == 0`, `title == ""` |
| a loop in a loop | `for record in records:` / `for role, name in record["repTeam"]:` |
| own rows | `row = {…}`, `row["critic"] = …`, `rows.append(row)`, `pd.DataFrame(rows)` |
| asking a table | `value_counts()`, `df[df["year"] >= 2015]`, `sort_values()`, `isna()`, `plot()` |
| cleaning | `replace({…})`, `.str.strip(".")`, `.str.contains("Bühne")`, `.str.endswith(".")` |
| linking | `team.merge(reviews[[…]], on="article_id")` |
