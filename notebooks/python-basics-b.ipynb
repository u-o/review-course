### 1. Lists

A **list** holds several values in a fixed order.

It is written with square brackets, and the values (we call them **items**) are separated by commas

````python
people = ["Bertolt Brecht", "Erich Engel", "Helene Weigel", "Angelika Hurwicz", "Paul Dessau"]
print(people)
print(type(people))
````

````python
# len() gives the number of items
print(len(people))
````

````python
# square brackets pick an item by its position, as with the characters of a string; counting starts at 0
print(people[0])
print(people[2])
print(people[-1])
````

````python
# there are five items, at the positions 0 to 4; position 5 does not exist: IndexError
print(people[5])
````

````python
# slices work as with strings: from position 1 up to (but not including) 3
print(people[1:3])
print(people[:2])
````

````python
# in checks whether an item is in the list; only exact matches count
print("Helene Weigel" in people)
print("Weigel" in people)

# compare: in a string, a part of the text is enough
print("Weigel" in "Helene Weigel")
````

````python
# a list can also hold numbers: the years of some of Brecht's premieres
years = [1922, 1923, 1928, 1941, 1943, 1948]
print(len(years))
print(min(years), max(years))
````

````python
# append() adds an item at the end
# unlike the string methods, append() changes the list itself
people.append("Werner Hinz")
print(people)
````

````python
# often we start with an empty list and fill it step by step
cast = []
print(cast)
cast.append("Helene Weigel")
cast.append("Paul Bildt")
print(cast)
print(len(cast))
````

### 2. From a text to a list of words

`split()` cuts a string into pieces and gives us a list.

This is our first step from a text to something we can count

````python
sentence = "Die Mutter Courage ist kein Klassenkampfstück"
words = sentence.split()
print(words)
print(len(words))
````

````python
# split() cuts at the spaces; punctuation stays attached to the word
words = "Aber alle Gestalten sind realistisch gesehen.".split()
print(words)
print("gesehen" in words)
print("gesehen." in words)
````

````python
# with an argument, split() cuts at another text
names = "Engel, Brecht, Weigel".split(", ")
print(names)
````

````python
# join() goes the other way: it glues the items of a list together, with the string before the dot in between
print(" ".join(names))
print(" + ".join(names))
````

````python
# "\n" is a line break: a character we cannot see, but which is there
text = "Erste Zeile\nZweite Zeile"
print(text)

# repr() shows the string the way Python sees it, with the line break as \n
print(repr(text))

# split() without an argument also cuts at line breaks
print(text.split())
````

### 3. for-loops

Often we want to do the same thing with every item of a list: print every name, measure every word, count every review.

A **for-loop** does this:

```python
for person in people:
    print(person)
```

Read it as: *take the first item of `people`, call it `person`, and run the indented lines; then take the second item, call it `person`, and run the indented lines again; and so on, until the list is used up*.


- the first line ends with a **colon**
- the lines that belong to the loop are **indented** (by four spaces; Colab does this for you after the colon)
- `person` is a new variable, and its name is our choice

````python
for person in people:
    print(person)
````

````python
# the name of the variable does not matter to Python, only to the humans reading the code
for x in people:
    print(x)
````

````python
# everything indented is repeated; the first line that is not indented runs once, after the loop
for person in people:
    print("on stage:", person)
    print("---")
print("end of the list")
````

````python
# without indentation Python does not know what belongs to the loop: IndentationError
for person in people:
print(person)
````

````python
# a loop over the words of a sentence; each word is printed with its length
words = "Die Mutter Courage ist kein Klassenkampfstück".split()
for word in words:
    print(word, len(word))
````

````python
# a loop over a string goes through it character by character
for letter in "Brecht":
    print(letter)
````

Two patterns we will need again and again: **counting** and **collecting**.

Look at what happens in each round of the loop

````python
# counting: a variable starts at 0 and grows by 1 in every round
count = 0
for word in words:
    count = count + 1
    print("round", count, "- the word is", word)
print("number of words:", count)
````

````python
# collecting: an empty list gets one new item in every round
lengths = []
for word in words:
    lengths.append(len(word))
    print(lengths)
print("all lengths:", lengths)
````

````python
# enumerate() gives two things in every round: the position and the item; we catch them in two variables
for number, person in enumerate(people):
    print(number, person)
````

### 4. Comparisons: True or False

A **comparison** gives back a truth value, `True` or `False` (the type `bool`)

| operator | meaning |
|---|---|
| `==` | equal (two equal signs: one `=` puts a value into a variable!) |
| `!=` | not equal |
| `<`, `>` | smaller, greater |
| `<=`, `>=` | smaller or equal, greater or equal |
| `in`, `not in` | is contained in a string or a list |

````python
year = 1941
print(year == 1941)
print(year != 1941)
print(year < 1933)
print(year >= 1933)
````

````python
# texts can be compared as well; upper and lower case make a difference
print("Brecht" == "Brecht")
print("Brecht" == "brecht")
print("Weigel" not in people)
````

````python
# and: both conditions must be True; or: one is enough; not turns True into False and back
print(year > 1933 and year < 1949)
print(year < 1933 or year > 1949)
print(not year < 1933)
````

### 5. if-clauses

With an **if-clause** the program decides: the indented lines are only run when the **condition** is `True`.

As with the for-loop: colon at the end, indentation below

````python
year = 1941
if year < 1933:
    print("premiered in the Weimar Republic")
print("done")
````

````python
# else catches everything else
if year < 1933:
    print("premiered in the Weimar Republic")
else:
    print("premiered after 1933")
````

````python
# elif (else if) adds further conditions; they are checked from top to bottom, and only the first True one counts
# change the year and run the cell again
year = 1941
if year < 1933:
    print("before the exile")
elif year < 1949:
    print("during the exile")
else:
    print("after the return to Berlin")
````

````python
sentence = "Und nun kommt Brechts Chronik vom großen Krieg."
if "Krieg" in sentence:
    print("the sentence is about war")
````

### 6. Loops and if-clauses together

Now the real work begins: go through every item, and decide for each one what to do.

We take a longer part of the review of 1949. The three quotation marks let a string run over several lines

````python
text = """Brecht ist in den zwanziger Jahren als der Auflöser des Dramas bekanntgeworden, als der kühne,
brutale Autor der Lehrstücke, der Präzeptor eines gnadenlos undichterischen Theaterstils. Indessen
aber hat diese Auflösung des Dramatischen die Bühnen der Welt erfüllt. Thornton Wilder und William
Saroyan sind gekommen, die transponierende Conférence und der Bilderbogen sind vertrauter „Stil“
geworden. Und nun kommt Brechts Chronik vom großen Krieg. Doch das ist kein Bilderbogen mehr und
auch kein Anklage- und Thesenstück, wie sie sich zu Dutzenden aus dem mißverstandenen Brecht
entwickelt haben. Die „Mutter Courage“ ist kein Klassenkampfstück. Aber alle Gestalten sind
realistisch gesehen. Es gibt in dieser Chronik keine Menschen, die restlos gut, und keine, die
restlos böse sind. Menschen sind sie, ausgeliefert dem Kriege."""

words = text.split()
print(len(words))
print(words[:12])
````

````python
# print only the long words
for word in words:
    if len(word) > 12:
        print(word)
````

````python
# counting with a condition: how many words contain "Brecht"?
count = 0
for word in words:
    if "Brecht" in word:
        count = count + 1
print(count)
````

````python
# collecting with a condition: all words that start with a capital letter (in German: nouns, names, and the first word of a sentence)
# isupper() checks whether a character is upper case
capitals = []
for word in words:
    if word[0].isupper():
        capitals.append(word)
print(len(capitals))
print(capitals)
````

````python
# the punctuation is in the way: "Menschen" and "Menschen," are different items
print(words.count("Menschen"))

# strip() removes the given characters at the start and the end of a string
print("„Stil“".strip("„“.,"))

# so we collect clean words first, and count again
clean = []
for word in words:
    clean.append(word.strip(".,;:„“"))
print(clean.count("Menschen"))
````

````python
# a loop over numbers with if, elif and else
years = [1922, 1923, 1926, 1928, 1930, 1936, 1941, 1943, 1948, 1956, 1958]
for year in years:
    if year < 1933:
        print(year, "before the exile")
    elif year < 1949:
        print(year, "during the exile")
    else:
        print(year, "after the return")
````

````python
# a loop inside a loop: for every sentence we go through its words
# first we remove the line breaks: split() cuts at them, join() glues the pieces together with spaces
# then split(". ") cuts the text at every full stop followed by a space
text = " ".join(text.split())
sentences = text.split(". ")
for sentence in sentences:
    capitals = 0
    for word in sentence.split():
        if word[0].isupper():
            capitals = capitals + 1
    print(capitals, "|", sentence[:60])
````
