# 05 – The NLP pipeline

In the last session we worked with what the archive says *about* the reviews: titles, dates, names of directors. Today we look at the text itself. Natural Language Processing (NLP) is the name for the methods that let a computer find structure in language: words, sentences, parts of speech, names. These steps run one after another, each adding information to the text, which is why one speaks of a pipeline

We use spaCy, the most common NLP library in python, and a single, older review: the premiere of Brecht's "Mutter Courage" in Berlin in 1949. We go through the pipeline step by step and look at what comes out of each step, and at what goes wrong

**Before we start**

The spaCy library is already installed in Colab.

spaCy uses *models* to process language. A model is a file that contains what a machine learning algorithm has learnt from thousands of sentences in which people marked by hand the type of words and the structure of the sentences. From these examples it has learnt to make assumptions about texts it has never seen. 

There are models for many languages and in several sizes.

We download the small German one, `de_core_news_sm`: `de` for German, `core` for the general tasks, `news` because it was trained mainly on newspaper texts, `sm` for small.

### 2.

The first step of every pipeline: the text is cut into words and sentences. We have just seen a simple way to do this: `split()` cut the text into pieces. Are these pieces words?

```python
# the same as above, now under the name words
words = plaintext.split()
print(len(words))
print(words[:40])
```

This almost works, but the punctuation sticks to the words: `„Deutsche`, `Theater“`, `gehabt.` are not words. And when we count, `Brecht` and `Brecht.` would be two different words. 
A **tokenizer** knows the rules of a language: it separates punctuation, but keeps abbreviations like "z.B." or dates like "11.1.1949" together. 
In NLP, the pieces are called **tokens** rather than words, because a comma is a token too.

```python
# spacy.load() loads the model; nlp is now the whole pipeline
import spacy

nlp = spacy.load("de_core_news_sm")

# pipe_names lists the steps that follow the tokenizer
print(nlp.pipe_names)
```

What the steps do:

- `tok2vec` turns every token into a list of numbers that the other steps work with
- `tagger` and `morphologizer` determine the part of speech (noun, verb, ...) and grammatical features (case, number, tense, ...)
- `parser` analyses the structure of each sentence: which word depends on which; it also decides where a sentence ends
- `lemmatizer` finds the base form of each word: "Kriege" → "Krieg"
- `attribute_ruler` corrects the results of the others with a few fixed rules
- `ner` (named entity recognition) finds names of persons, places and organisations

Except for the rules, none of this is programmed by hand. The model has learnt from German newspaper texts of the 1990s and from Wikipedia articles, annotated by people, and it makes predictions about our text based on them. Predictions can be wrong, and a review of 1949 is neither a newspaper of the 1990s nor a Wikipedia article

```python
# we run the whole pipeline over the text with one call; the result is a Doc
doc = nlp(plaintext)
print(type(doc))
```

```python
# a Doc is basically like a list of tokens, len returns the length of the list 
print(len(doc))
# so doc[x] returns the xth token, with .text we get the token itself, .i returns the position of the token in the tex
print(doc[0].i, doc[0].text)
```

```python
# loop over the first 40 tokens and print their position and text

for token in doc[:40]:
    print(token.i, token.text)
```

```python
# doc.sents gives the sentences; list() turns them into a list that we can count and slice
sentences = list(doc.sents)
print(len(sentences))

for sentence in sentences[:10]:
    print(sentence.text)
    print("---")
```

Mostly right, but "Thornton" stands alone as a sentence, and the next one begins with "Wilder". The model did not recognise the American playwright Thornton Wilder and read "Wilder" as the adjective "wilder" at the beginning of a new sentence. Splitting sentences is not just looking for full stops: the parser decides, and the parser makes mistakes

### 4.

The morphologizer adds grammatical features to each token: case, gender, number, and for verbs person, tense and form

```python
for token in doc[:12]:
    print(token.text, token.pos_, token.morph)
```

```python
# morph.get() gives the value of one feature, as a list (it is empty if the token has no tense)
for token in doc[:60]:
    if token.pos_ in ["VERB", "AUX"]:
        print(token.text, token.morph.get("Tense"), token.morph.get("VerbForm"))
```

"hat ... gehabt", "hat ... gemacht": in German, the past is often told in the perfect tense, an auxiliary in the present ("hat") and a participle (`VerbForm=Part`). The model only gives a tense to the auxiliary, so for the computer "hat gehabt" is present. We therefore count three things in every sentence: forms in the simple past (`Past`), in the present (`Pres`), and participles

```python
rows = []
for number, sentence in enumerate(sentences):
    past = 0
    present = 0
    participles = 0
    for token in sentence:
        if "Past" in token.morph.get("Tense"):
            past = past + 1
        if "Pres" in token.morph.get("Tense"):
            present = present + 1
        if "Part" in token.morph.get("VerbForm"):
            participles = participles + 1
    rows.append({"sentence": number, "past": past, "present": present, "participles": participles, "text": sentence.text})

tenses = pd.DataFrame(rows)

# pd.set_option() shows the whole text in each cell
pd.set_option("display.max_colwidth", None)
tenses
```

The text has three layers of time, and the grammar separates them:

- the **perfect** (participles, most of sentences 0–8) tells the history of the institution and of the author: since Langhoff's Intendanz the stage "hat keine großen Abende mehr gehabt", Brecht "ist in den zwanziger Jahren ... bekanntgeworden", Wilder and Saroyan "sind gekommen"
- the **present** describes the play: what it is and what happens in it
- the **simple past** (sentences 18–19) tells the evening itself: "So wurde dies Stück ... gespielt", "Es war eine dichterische Verklärung"

In the review of 2013 from the nachtkritik archive, the simple past told the history and the present the evening. The 1949 critic uses the simple past for the performance, a convention of the newspaper review at the time. The same feature means different things in different texts. And not every participle is a perfect: "sind realistisch gesehen" and "wurde gespielt" are passives

### 5.

The parser analyses the structure of each sentence. Every token gets a head, the word it depends on, and a label that names the relation. The labels of the German model come from the TIGER corpus: `sb` is the subject, `oa` the accusative object, `ag` a genitive attribute, `app` an apposition, `oc` a dependent verb

```python
# the second sentence: the return of Brecht
sentence = sentences[1]
print(sentence.text)

for token in sentence:
    print(token.text, token.dep_, token.head.text)
```

How to read this output: the parser turns every sentence into a tree. We go through it from the top down, using our sentence

> Die Wiederkehr Brechts und die Entdeckung eines neuen, wesentlich gewandelten Brecht in der „Mutter Courage“ jedoch hat den Namen der Bühne wieder lebendig gemacht.

**1. The root.** Every sentence has exactly one token that depends on nothing: the `ROOT`, usually the conjugated verb. Here it is "hat". All other tokens depend on the root, directly or through other tokens. (English: in "The return of Brecht has revived the stage", the root is "has")

**2. The verb complex.** In German, a verb is often split into two parts. The participle depends on the auxiliary as a dependent verb (`oc`): "gemacht → hat". A separable prefix is marked `svp`: "er zieht sich etwas hin" gives "hin → zieht" (English has something similar in phrasal verbs: "the war drags on")

**3. The participants.** The verb has its participants:

- the subject (`sb`), who or what does something: "Wiederkehr → hat" (English: "return → has")
- the accusative object (`oa`), what is affected: "Namen → gemacht" (English: "made the name alive")
- the dative object (`da`), to or for whom: "den Marketendern ... gilt Brechts Vorliebe" gives "Marketendern → gilt" (English: "gave him the book")
- the predicate (`pd`) after "sein" or "werden": "ist kein Klassenkampfstück" gives "Klassenkampfstück → ist" (English: "is no play of class struggle")

Note that the subject depends on "hat", the object on "gemacht": the tree follows the split verb

**4. The circumstances.** Words and phrases that say how, when, where or why are modifiers (`mo`): "wieder → gemacht", "lebendig → gemacht" (English: "made ... alive again"). "seit Langhoffs Intendanz" in the first sentence is a modifier too. Negation has a label of its own (`ng`): "nicht → dienen". Here the parser also attached "jedoch" to "Wiederkehr" instead of to the verb: a small error

**5. Inside the noun phrases.** A noun is the head of its own small tree:

- articles and adjectives belong to it as noun kernel elements (`nk`): "Die → Wiederkehr", "den → Namen" (English: "the → return"). After a preposition, the noun itself is `nk` of the preposition: "Courage → in"
- a genitive is a genitive attribute (`ag`): "Brechts → Wiederkehr", "Bühne → Namen" (English: "Brecht's return", "the name of the stage")
- a prepositional phrase that belongs to a noun is `mnr`: "in → Brecht" ("der Brecht in der Mutter Courage"; English: "the Brecht in Mother Courage")
- the parts of a name are `pnc`: "Mutter → Courage", "Helene → Weigel"
- an apposition (`app`) explains a noun: "Helene Weigel, der Frau des Dichters" gives "Frau → Weigel" (English: "Helene Weigel, the poet's wife")

**6. Coordination.** "und" and "oder" are coordinating conjunctions (`cd`) and depend on the first part; the second part depends on "und" as a conjunct (`cj`): "und → Wiederkehr", "Entdeckung → und" (English: "the return and the discovery")

**7. Clauses and small words.** A relative clause depends on the noun it describes (`rc`): "des Dramas, das mit der Strophe schließt" gives "schließt → Dramas". Words that open a subordinate clause are `cp` ("während", "um", "ohne"), "zu" before an infinitive is `pm`, "als" and "wie" in comparisons are `cm`, a conjunction at the beginning of a sentence ("Doch", "Aber") is `ju`, and an "es" without meaning of its own ("Es gibt ...") is `ep`

**8. Punctuation** (`punct`) usually depends on the root or on the word it closes

`spacy.explain()` gives the English name of each label, e.g. `spacy.explain("nk")`. In the drawing below, each arc runs from a head to the token that depends on it, and carries the label

```python
# displacy draws the structure as a tree of arcs
from spacy import displacy

displacy.render(sentence, style="dep", jupyter=True, options={"compact": True, "distance": 100})
```

The subject is "Wiederkehr" (`sb`), and "Brechts" depends on it as a genitive (`ag`): the return of Brecht. But the subject hangs on the auxiliary "hat", while the object "Namen" hangs on the participle "gemacht" (`oc`). The perfect splits the verb into two, and the tree splits with it

With subject, verb and object we can ask: who does what? For every subject we take its verb; if that is an auxiliary with a dependent verb (`oc`), we take the dependent verb instead. Then we collect the accusative objects (`children` are the tokens that depend on a token)

```python
for token in doc:
    if token.dep_ == "sb":
        verb = token.head
        for child in token.head.children:
            if child.dep_ == "oc" and child.pos_ == "VERB":
                verb = child
        objects = []
        for child in verb.children:
            if child.dep_ == "oa":
                objects.append(child.text)
        print(token.text, "|", verb.lemma_, "|", objects)
```

The first lines read like a summary of institutional history: "Theater – haben – Abende" (the Deutsches Theater has had no great evenings), "Wiederkehr – machen – Namen" (the return of Brecht made the name of the stage alive again), "Auflösung – erfüllen – Bühnen" (the dissolution of drama filled the stages of the world). Many subjects are pronouns ("die", "er", "sie"), and some lines are wrong: "Song – durchklingen – Wanderung" reverses the sentence, and in "Trommelwirbel – wissen" our rule took the wrong verb, because the tagger took "übertönt" for a participle

The parser also helps with time. Expressions like "in den zwanziger Jahren" are a preposition and the words that depend on it; `subtree` gives us a token together with everything that depends on it

```python
time_words = ["Intendanz", "Jahr", "Beginn", "Nachfolge"]

for token in doc:
    if token.lemma_ in time_words and token.head.pos_ == "ADP":
        phrase = []
        for word in token.head.subtree:
            phrase.append(word.text)
        print(" ".join(phrase))
```

"seit Langhoffs Intendanz und der sich damit anbahnenden Ost-Politisierung dieser berühmten Bühne", "in den zwanziger Jahren", "in der Nachfolge Kurt Weills", "zum thesenhaften Beginn Brechts": the text measures time not in dates but in Intendanzen, decades, and in the careers of artists

### 6.

The last step: named entities. The German model knows four kinds: persons (`PER`), places (`LOC`), organisations (`ORG`) and everything else (`MISC`)

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# displacy marks the entities in the text
displacy.render(doc, style="ent", jupyter=True)
```

```python
rows = []
for ent in doc.ents:
    rows.append({"text": ent.text, "lemma": ent.lemma_, "label": ent.label_})

ents = pd.DataFrame(rows)
ents["label"].value_counts()
```

```python
# the persons; we count the lemmas, so that "Brechts" counts as "Brecht"
ents[ents["label"] == "PER"]["lemma"].value_counts()
```

Most persons are found: Langhoff, Brecht, Engel, Weigel, Hurwicz, Hinz, Bildt, Dessau, Weill. But "Thornton" is found without "Wilder" (see above), "Tillys" (the general Tilly) is missed, "Paul Dessaus" keeps its genitive "s", and "Dramas", "Conférence" and "Kriegsmythus" are taken for persons. "Brechts" is once a place, and "Krieg" is marked as an entity again and again, because in Wikipedia wars are names ("Dreißigjähriger Krieg"). And the one institution the text is about, the "Deutsche Theater", is not found as an organisation: only "Deutsche" is marked. The only `ORG` is "Nachfolge". For questions about institutions, this is the weakest step of the pipeline

### 7.

What can the pipeline tell us about institutional change? A few simple questions, each combining the steps from above

**Which persons stand for which period or position?** In this text, persons often appear in the genitive: "Langhoffs Intendanz", "die Wiederkehr Brechts". The parser marks this relation with `ag`

```python
for token in doc:
    if token.dep_ == "ag" and token.pos_ == "PROPN":
        print(token.text, "->", token.head.text)
```

"Langhoffs Intendanz", "Wiederkehr Brechts", "Nachfolge Kurt Weills", "Beginn Brechts": a small vocabulary of institutional time, in which persons stand for periods. One is missing: "die Entdeckung eines neuen ... Brecht", because the tagger took this "Brecht" for a noun (step 3)

**Who has which role?** In this text the role follows the name, separated by a comma: "Helene Weigel, der Frau des Dichters". The parser calls this an apposition (`app`)

```python
for token in doc:
    if token.dep_ == "app":
        phrase = []
        for word in token.subtree:
            phrase.append(word.text)
        print(token.head.text, "|", " ".join(phrase))
```

"Weigel – der Frau des Dichters", "Hurwicz – der stummen und vielfältig geschlagenen Tochter", "Hinz – dem Feldprediger", "Bildt – dem Koch", and in the other direction "Regisseur – Erich Engel". The review mixes the roles in the play with the roles in life: Weigel is both Mutter Courage and Brecht's wife. Some lines are noise ("er – der Krieg ..."), and Bildt is wrongly attached to Hinz. In the 2013 review, the role stood before the name ("Kultursenator Moritz Thape"): each pattern finds only what it describes, and each text has its own patterns

**Which words of the institution occur, and where?** We search the lemmas for a short list of words. With `in` we also find compounds like "Ost-Politisierung" or "Theaterstil"

```python
institution_words = ["intendan", "theater", "bühne", "polit", "wiederkehr", "regisseur", "nachfolge", "epigon"]

for token in doc:
    for word in institution_words:
        if word in token.lemma_.lower():
            print(token.i, token.text, "|", token.sent.text[:120])
```

Put together, a short history emerges from the text: a famous stage that has had no great evenings since a new Intendant and the "Ost-Politisierung", an author who returns and is rediscovered, a director from the 1920s, an actress who is also the author's wife, a composer in the succession of another. All of this is in the text, and for a single review we could have found it by reading. But each step has shown us its errors, and we could only notice them because we had read the text. Next time we run the pipeline over hundreds of reviews, where nobody can read along any more
