### 2. Working with a title

The titles on nachtkritik.de follow a pattern: first the name of the play, then a dash `–`, then a short description of the evening.

Strings come with methods: `lower()`, `upper()`, `count()`, `replace()`. And `in` checks whether a text appears in a string

````python
print(len(title))
print(title.lower())
print(title.count("e"))
print("Graz" in title)
print("graz" in title)
````

````python
# replace() gives back a new string; title itself does not change
print(title.replace("Konwitschny", "K."))
print(title)
````

````python
# positions and slices
print(title[0])
print(title[:5])
print(title[-10:])
````

### 4. From a string to a list and back: split() and join()

`split()` cuts a string into a list. Without an argument it cuts at the spaces; with an argument it cuts at that text.

This is useful for the titles: if we cut at the dash ` – ` (with a space on each side), we get the play and the description apart

````python
parts = titles[3].split(" – ")
print(parts)
print(len(parts))
print(parts[0])
print(parts[1])
````

````python
# a name cut at the spaces: the first item is the first name, the last item the surname
name_parts = directors[3].split()
print(name_parts)
print(name_parts[-1])

# split() and the position can also be written in one go
print(directors[3].split()[-1])
````

````python
# a date cut at the hyphens
print(dates[3].split("-"))

# join() goes the other way: it glues the items of a list together, with the string before the dot in between
print(" / ".join(directors[:3]))
````

### 5. First loops

A for-loop does the same thing with every item of a list. Read it as: *take the first item of `titles`, call it `title`, run the indented lines; then take the second item, …*

- the first line ends with a colon `:`
- the lines that belong to the loop are indented

````python
# the name of the play: the part of every title before the first dash
for title in titles:
    print(title.split(" – ")[0])
````

````python
for director in directors:
    print(director, len(director))
````

````python
# counting: a variable starts at 0 and grows by 1 in every round
count = 0
for place in places:
    count = count + 1
print(count)
````

````python
# collecting: an empty list gets one new item in every round
years = []
for date in dates:
    years.append(int(date[:4]))
print(years)
print(min(years), max(years))
````
