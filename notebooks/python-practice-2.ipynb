### 3. Counting and collecting with a condition

The two patterns from *basics B* that we need again and again:

- **counting**: a variable starts at `0` before the loop and grows by `1` when the condition is true
- **collecting**: an empty list `[]` before the loop gets a new item when the condition is true

````python
# counting
count = 0
for place in places:
    if place == "Köln":
        count = count + 1
print(count)
````

````python
# collecting: the plays whose name is a single word
short_names = []
for title in titles:
    play = title.split(" – ")[0]
    if len(play.split()) == 1:
        short_names.append(play)
print(short_names)
````

### 4. More than two cases: if, elif, else

`elif` adds further conditions; they are checked from top to bottom, and only the first true one counts. `else` catches everything else

````python
for year in years:
    if year < 2010:
        print(year, "the early years")
    elif year < 2020:
        print(year, "the 2010s")
    else:
        print(year, "the 2020s")
````

### 5. Several lists in step: enumerate()

Our lists all have the same order. If a loop knows the **position** of the review, it can look up everything else about it in the other lists.

`enumerate()` gives two things in every round: the position and the item

````python
for i, date in enumerate(dates):
    print(i, date, critics[i])
````

````python
# which reviews have no critic in the export?
for i, critic in enumerate(critics):
    if critic == "":
        print(i, dates[i], titles[i])
````

### 6. The words of a review

The same patterns work on the words of a text. Here are the first three sentences of Petra Hallmayer's review of *Der Riss durch die Welt* at the Residenztheater München (2019)

````python
text = "Der Abend beginnt mit einem langen Schweigen. Stumm sitzen zwei Männer und zwei Frauen auf rotgepolsterten Stühlen auf der leeren dunklen Bühne. In Rückblenden und Zeitsprüngen erzählt Roland Schimmelpfennig in seinem im Auftrag des Bayerischen Staatsschauspiels entstandenen Stück „Der Riss durch die Welt“, das von Tilmann Köhler nun am Cuvilliéstheater uraufgeführt wurde, von einer radikal misslingenden Begegnung."

words = text.split()
print(len(words))
print(words[:10])
````

````python
# punctuation sticks to the words; strip() removes the given characters at the start and the end
clean = []
for word in words:
    clean.append(word.strip(".,„“"))
print(clean[:10])
print(clean[-5:])
````

### 7. Counting many things at once: a dictionary

With a counter variable we can count one thing. To count **many different things at once** (how many reviews per place, per critic, per year?) we use a dictionary: every thing becomes a key, and its number the value (*basics C*, section 2).

Watch the dictionary grow in every round

````python
counts = {}
for place in places:
    if place in counts:
        counts[place] = counts[place] + 1     # we have seen the place before: add 1
    else:
        counts[place] = 1                     # the place is new: start with 1
    print(place, "→", counts)
````

````python
print(counts)
print(counts["Köln"])

# a loop over a dictionary goes through its keys
for place in counts:
    print(place, counts[place])
````

### Summary

| concept | example |
|---|---|
| a loop with if | `for place in places:` / `if place == "Köln":` |
| counting | `count = 0` … `count = count + 1` |
| collecting | `found = []` … `found.append(title)` |
| several cases | `if …:` / `elif …:` / `else:` |
| several lists in step | `for i, date in enumerate(dates):` … `critics[i]` |
| cleaning words | `word.strip(".,„“")`, `word.lower()`, `word[0].isupper()`, `word.endswith("en")` |
| counting with a dictionary | `if place in counts: … else: …` |

Keeping every kind of information in a list of its own works, but it is fragile: one missing item, and all positions after it are wrong. That is why real data keep everything about one review together, in a dictionary.
