## 1st lesson: commands, variables, types

### 1. Python

A **program** consists of lines of **code**. Each line is a **command** for the computer.

The commands are carried out one after the other, from top to bottom. 

A program basically reads: do this, then do that, then do that.

A very simple program looks like this:

```python
print("hello world")
```

It tells the computer to show (print) the text `hello world`.

### 5. Functions and arguments

`print` is a **function**. 

A function is a command with a name that does a certain job. We call it by writing its name followed by round brackets. What we put inside the brackets is the **argument**: the value the function works with.

```
print("hello world")
  ↑        ↑
function   argument
```

- brackets always come in pairs: every `(` needs its `)`
- text always stands between quotation marks: `"hello world"`. Numbers are written without them

````python
print("Brecht")
print(1898)
````

````python
# a function can take several arguments, separated by commas; print puts a space between them
print("Brecht was born in", 1898)
````

````python
# some arguments have a name; sep= says what print puts between the other arguments (sep stands for separator)
print("Brecht", "Weill", "Neher")
print("Brecht", "Weill", "Neher", sep=" / ")
````

````python
# len() is another function: it gives back the length of a text, i.e. the number of its characters (spaces included)
# to see the result, we hand it on to print: the inner function is run first, then the outer one
print(len("Mutter Courage"))
````

````python
# in a notebook the result of the LAST line of a cell is shown even without print
len("Mutter Courage")
````

### 6. Variables

Most of the time we work with data that changes: a different review, a different year, a different name. 

That is why we need containers into which we can put data, take it out again, and put something else in. Such a container is called a **variable**.

A variable is like a jar with a label: the label is its name, and through the name we get at what is inside. We fill it with the equal sign `=`

````python
# put the text "Monty Jacobs" into a variable named critic
critic = "Monty Jacobs"

# from now on we can use the name instead of the value
print(critic)
````

Careful: `=` does not mean "is equal" as in mathematics. It means: **put the value on the right into the variable on the left**. When a cell only assigns variables, it shows no output; the values are kept in memory

````python
# the names are up to us; the computer does not care what they mean, they are for the humans who read the code
a = 5                                       # short, but the name says nothing about what is inside
number_of_reviews_in_our_collection = 5     # says everything, but is long to type
reviews = 5                                 # a good compromise
````

Rules for names:

- letters, digits and the underscore `_`, no spaces, no hyphens
- a name must not start with a digit
- upper and lower case make a difference: `Critic` and `critic` are two different variables
- we write names in lower case, with underscores between the words: `year_of_birth`

````python
# a variable can get a new value at any time; the old value is gone
critic = "Monty Jacobs"
print(critic)

critic = "Alfred Kerr"
print(critic)
````

````python
# a variable can be used to calculate a new value
year = 1928
next_year = year + 1
print(next_year)
````

````python
# this line looks strange in mathematics, but makes sense in Python: read it from right to left
# take the value of count, add 1, and put the result back into count
count = 0
count = count + 1
count = count + 1
count = count + 1
print(count)
````

### 7. Types

Variables can contain different kinds of values. Every value has a **type**. For now we need four of them:

| type | name in Python | examples |
|---|---|---|
| text (a *string* of characters) | `str` | `"Brecht"`, `"1898"`, `""` |
| whole number (*integer*) | `int` | `1898`, `0`, `-3` |
| decimal number (*float*) | `float` | `3.5`, `0.25` (with a point, not a comma) |
| truth value (*boolean*) | `bool` | `True`, `False` |

The quotation marks make the difference: whatever stands between them is text

````python
year_of_birth = 1898        # a whole number: type int
first_name = "Bertolt"      # a text: type str, recognised by the quotation marks
running_time = 3.5          # a decimal number: type float ("dreieinhalbstündig", says the review of 1949)
in_exile = True             # a truth value: type bool (more about this in the next notebook)

# type() tells us the type of a value
print(type(year_of_birth))
print(type(first_name))
print(type(running_time))
print(type(in_exile))
````

````python
# a number between quotation marks is text, not a number
year_as_text = "1898"
print(type(year_as_text))
````

What we can do with a value depends on its type. `+` adds numbers, but glues texts together

````python
print(2 + 3)
print("2" + "3")
````

````python
first_name = "Bertolt"
last_name = "Brecht"

# texts are glued together; spaces and commas have to be added by ourselves
name = last_name + ", " + first_name
print(name)

year_of_birth = 1898
year_of_death = 1956

# numbers are calculated: + - * / work as usual
age = year_of_death - year_of_birth
print(age)
````

````python
# text and numbers cannot simply be glued together: Python answers with a TypeError
print("Brecht died at the age of " + age)
````

````python
# str() turns a number into a text, and the mistake is solved
print("Brecht died at the age of " + str(age))

# int() goes the other way and turns a text into a number
print(int("1898") + 1)
````

````python
# a shortcut: print accepts arguments of different types, separated by commas, and adds the spaces itself
print("The author", name, "lived for", age, "years")
print("The author", first_name, last_name, "lived for", year_of_death - year_of_birth, "years")
````

### 8. Working with text

Our data are texts, so we spend a lot of time with strings. Here is a sentence from a review of Brecht's *Mutter Courage* at the Deutsches Theater Berlin, 1949

````python
sentence = "Die Wiederkehr Brechts und die Entdeckung eines neuen, wesentlich gewandelten Brecht in der „Mutter Courage“ jedoch hat den Namen der Bühne wieder lebendig gemacht."
print(sentence)
print(len(sentence))
````

Strings come with functions of their own, called **methods**. 

A method is written after the value (or the variable), with a dot: `sentence.lower()`. The brackets are needed even when there is no argument

````python
# lower() and upper() give us the text in lower or upper case
print(sentence.lower())
print(sentence.upper())
````

````python
# count() counts how often a text appears in the string
print(sentence.count("Brecht"))
print(sentence.count("die"))
````

````python
# "die" was found only once: upper and lower case make a difference ("Die" at the start was not counted)
# methods can be chained: first lower(), then count() on the result
print(sentence.lower().count("die"))
````

````python
# replace() swaps one text for another
print(sentence.replace("Brecht", "B."))
````

````python
# methods do not change the variable itself: sentence is still the same
print(sentence)

# to keep the result, we put it into a variable
short = sentence.replace("Brecht", "B.")
print(short)
````

````python
# in checks whether a text appears in the string; the answer is a truth value (bool)
print("Courage" in sentence)
print("Weigel" in sentence)
````

With square brackets we pick single characters or pieces of a string by their position. **Counting starts at 0**, not at 1

````python
print(sentence[0])      # the first character
print(sentence[1])      # the second
print(sentence[-1])     # negative positions count from the end: the last character
````

````python
# with a colon we get a slice: from position 4 up to (but not including) position 14
print(sentence[4:14])

# without a number before the colon: from the beginning; without a number after it: to the end
print(sentence[:14])
print(sentence[130:])
````

### 9. Mistakes

Programming is a lot about making mistakes and finding them. Error messages look frightening, but they help. 

**Read the last line first**: it names the kind of mistake, and above it an arrow often points to the place

````python
# a bracket is missing
print("hello world"
````

````python
# upper and lower case: the function is called print, not Print
Print("hello world")
````

The most common mistakes at the start:

| message | what it usually means |
|---|---|
| `SyntaxError` | a bracket or a quotation mark is missing, or a typo in the code itself |
| `NameError` | a variable or function with this name does not exist: a typo, or the cell that creates it has not been run |
| `TypeError` | a value of the wrong type, e.g. text + number |

Python finds some mistakes, but not all: when we type a wrong year, the result is wrong, but Python does not notice.
