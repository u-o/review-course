```python
# an integer is a whole number, written without quotation marks
year = 1928
print(year)
```

```python
# type() tells us the type of a value
print(type(review))
print(type(year))
```

### 2.

Strings can be measured and cut into pieces

```python
# len() gives the length of a string, that is the number of its characters (spaces and punctuation included)
print(len(play))
print(len(review))
```

```python
# square brackets pick a single character by its position; counting starts at 0, not at 1
print(play[0])
print(play[1])
```

```python
# negative positions count from the end
print(play[-1])
```

```python
# with a colon we get a slice, a part of the string: [6:] means from position 6 to the end
print(play[6:])
```

```python
# [:5] means from the beginning up to (but not including) position 5
print(play[:5])
print(review[:100])
```

```python
# find() gives the position where a text first appears in the string; with a slice we can look at what follows
position = review.find("Brecht")
print(position)
print(review[position:position + 60])
```

### 3.

With integers we can calculate

```python
# Brecht was born in 1898: how old was he at the premiere?
brecht_born = 1898
print(year - brecht_born)
```

```python
# / always gives a decimal number (a float, yet another type); // drops the remainder
# a standard typed page has 1800 characters: how many pages is our review?
print(len(review) / 1800)
print(len(review) // 1800)
```

```python
# + does different things to integers and strings: integers are added, strings are joined
print(2 + 3)
print("2" + "3")
print(play + " (" + author + ")")
```

```python
# programming is a lot about making mistakes: strings and integers cannot simply be mixed
print("the review has " + len(review) + " characters")
```

```python
# str() turns an integer into a string, and the mistake is solved
print("the review has " + str(len(review)) + " characters")
```

### 4.

Comparisons produce booleans, values that are either True or False. With an if-clause the program decides what to do

```python
# == compares two values (unlike =, which assigns a value)
print(year == 1928)
print(critic == "Alfred Kerr")
print(len(review) > 1000)
```

```python
# in checks whether a text appears somewhere in a string
print("Brecht" in review)
print("Weill" in review)
```

```python
# True and False are a type of their own
print(type(True))
```

```python
# if executes the indented lines only when the condition is True
if "Brecht" in review:
    print("the review mentions Brecht")
```

```python
# else catches everything else
if "Weill" in review:
    print("the review mentions the composer")
else:
    print("the review does not mention the composer")
```

```python
# and: both conditions must be True (python writes it in lowercase)
print("Brecht" in review and "Weill" in review)
```

```python
# or: one True condition is enough
print("Brecht" in review or "Weill" in review)
```

```python
# and and or can be used in if-clauses as well
if "Engel" in review and "Neher" in review:
    print("the review mentions the director and the stage designer")

if "Piscator" in review or "Reinhardt" in review:
    print("the review compares the production with another director")
```
