# 01 – First Steps

A first attempt at reading a text "from a distance": instead of reading it sentence by sentence, we count its words.

Run the cells one after another (`Shift` + `Enter`).

## 1. A text

A short, made-up example text in the style of a theatre review.

```python
text = """
The stage is empty, only a chair stands in the light. The production relies
entirely on the acting: the actress carries the evening, the direction holds
back. At the end the chair is still in the light, and the audience applauds at length.
"""

print(text)
```

## 2. Splitting the text into words

We lowercase everything, drop the punctuation and keep only the words.

```python
import re

words = re.findall(r"\w+", text.lower())

print("Number of words:", len(words))
print("Distinct words:", len(set(words)))
print(words[:10])
```

## 3. Which words occur most often?

```python
from collections import Counter

frequencies = Counter(words)

for word, count in frequencies.most_common(10):
    print(f"{word:15} {count}")
```
