# 03 – Getting the Course Files

This notebook copies the course files into your own Google Drive, into the folder `My Drive/review-course/data/`. They stay there after Colab closes, and later notebooks read them from there.

Choose **Runtime → Run all**. Google will ask for permission to access your Drive – allow it.

You can run this notebook again whenever new files are added. Files you already have are not overwritten.

## 1. Connect your Google Drive

````python
from pathlib import Path

from google.colab import drive

drive.mount("/content/drive")

DATA_DIR = Path("/content/drive/MyDrive/review-course/data")
````

## 2. Download the files

We download the whole course repository from GitHub as one zip file and copy only the folder `data/public/` into your Drive.

````python
import io
import urllib.request
import zipfile

REPO_ZIP = "https://github.com/u-o/review-course/archive/refs/heads/main.zip"
FOLDER_IN_ZIP = "review-course-main/data/public/"

with urllib.request.urlopen(REPO_ZIP) as response:
    archive = zipfile.ZipFile(io.BytesIO(response.read()))

new, existing = 0, 0
for name in archive.namelist():
    if not name.startswith(FOLDER_IN_ZIP) or name.endswith("/"):
        continue
    target = DATA_DIR / name[len(FOLDER_IN_ZIP):]
    if target.exists():
        existing += 1
        continue
    target.parent.mkdir(parents=True, exist_ok=True)
    target.write_bytes(archive.read(name))
    new += 1

print(f"{new} new files copied, {existing} already there.")
````

## 3. What's in your Drive now?

````python
for path in sorted(DATA_DIR.rglob("*")):
    if path.is_file():
        print(path.relative_to(DATA_DIR))
````
