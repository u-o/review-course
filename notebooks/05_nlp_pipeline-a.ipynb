# 05 – The NLP pipeline

In the last session we worked with what the archive says *about* the reviews: titles, dates, names of directors. Today we look at the text itself. Natural Language Processing (NLP) is the name for the methods that let a computer find structure in language: words, sentences, parts of speech, names. These steps run one after another, each adding information to the text, which is why one speaks of a pipeline

We use spaCy, the most common NLP library in python, and a single review from the nachtkritik archive. We go through the pipeline step by step and look at what comes out of each step, and at what goes wrong

### 2.

The first step of every pipeline: the text is cut into words and sentences. We already know a simple way to do this

```python
# split() cuts the text at every space
words = plaintext.split()
print(len(words))
print(words[:40])
```

This almost works, but the punctuation sticks to the words: `aufstieg,` and `Stil",` are not words. And when we count, `Bremen` and `Bremen.` would be two different words. Where a link was removed in the export, a space stands before the punctuation (`Fassbinder ,`), so here the comma became a word of its own. A tokenizer knows the rules of a language: it separates punctuation, but keeps abbreviations like "z.B." or dates like "3.6.2013" together. In NLP, the pieces are called tokens rather than words, because a comma is a token too

```python
# spacy.load() loads the model; nlp is now the whole pipeline
import spacy

nlp = spacy.load("de_core_news_sm")

# pipe_names lists the steps that follow the tokenizer
print(nlp.pipe_names)
```

What the steps do:

- `tok2vec` turns every token into a list of numbers that the other steps work with
- `tagger` and `morphologizer` determine the part of speech (noun, verb, ...) and grammatical features (case, number, tense, ...)
- `parser` analyses the structure of each sentence: which word depends on which; it also decides where a sentence ends
- `lemmatizer` finds the base form of each word: "Intendanten" → "Intendant"
- `attribute_ruler` corrects the results of the others with a few fixed rules
- `ner` (named entity recognition) finds names of persons, places and organisations

Except for the rules, none of this is programmed by hand. The model has learnt from German newspaper texts and Wikipedia articles that were annotated by people, and it makes predictions about our text based on them. Predictions can be wrong, and a review of 2013 is not a newspaper of the 1990s

```python
# we run the whole pipeline over the text with one call; the result is a Doc
doc = nlp(plaintext)
print(type(doc))

# a Doc behaves like a list of tokens
print(len(doc))
```

```python
# token.i is the position of the token in the text, token.text the token itself
for token in doc[:40]:
    print(token.i, token.text)
```

```python
# doc.sents gives the sentences; list() turns them into a list that we can count and slice
sentences = list(doc.sents)
print(len(sentences))

for sentence in sentences[:15]:
    print(sentence.text)
    print("---")
```

Mostly right, with two kinds of problems that come from the archive, not from spaCy. The intermediate headings of the review ("Der Vermieter von Bruno Ganz") lost their line break in the export; spaCy separates this one correctly, but not every heading. And the title "War da was?" ends with a question mark, so it was cut into two sentences

### 3.

Each token now carries information, which we get with attributes: `pos_` is the part of speech, `lemma_` the base form (the underscore gives us the readable text instead of a number)

```python
for token in doc[:25]:
    print(token.text, token.pos_, token.lemma_)
```

```python
# spacy.explain() explains the abbreviations
print(spacy.explain("PROPN"))
print(spacy.explain("ADP"))
print(spacy.explain("SCONJ"))
```

```python
# as in the last session, a list of dictionaries becomes a table: one row per token
import pandas as pd

rows = []
for token in doc:
    rows.append({
        "i": token.i,
        "text": token.text,
        "pos": token.pos_,
        "lemma": token.lemma_,
        "is_stop": token.is_stop,    # very frequent words like "der", "und", "ist"
        "is_punct": token.is_punct,  # punctuation
    })

tokens = pd.DataFrame(rows)
tokens.head(20)
```

```python
tokens["pos"].value_counts()
```

```python
# which nouns are most frequent? we count the lemmas, so that "Theater" and "Theaters" count as one
nouns = tokens[tokens["pos"] == "NOUN"]
nouns["lemma"].value_counts().head(20)
```

Theater, Abend, Jahr, Mythos, Schauspieler, Bühne, Spielzeit: even this crude list says what the text is about. Without lemmas, "Intendanten" and "Intendant" would be two different words, and so would "Theater" and "Theaters"

But the lemmatizer of the small model mostly looks the forms up in a list, and where a form is missing it guesses. And the tagger takes some names for nouns

```python
# isin() keeps the rows whose value is in the list
tokens[tokens["text"].isin(["aufstieg", "herausfand", "Erbes", "Fassbinder", "Zadek", "Ganz"])]
```

"aufstieg" should be "aufsteigen", "herausfand" "herausfinden", "Erbes" "Erbe"; Fassbinder and Zadek are names (PROPN), and the first "Ganz" (Bruno Ganz) gets the lemma "ganz". Every later step builds on these results, so errors are passed on through the pipeline

### 4.

The morphologizer adds grammatical features to each token: case, gender, number, and for verbs person, tense and mood

```python
for token in doc[:12]:
    print(token.text, token.pos_, token.morph)
```

```python
# morph.get() gives the value of one feature, as a list (it is empty if the token has no tense)
for token in doc[:30]:
    if token.pos_ in ["VERB", "AUX"]:
        print(token.text, token.morph.get("Tense"))
```

Tense is interesting for us. A review describes the evening in the present tense; when it speaks in the past tense, it often tells a story: here, the history of the institution. We count past and present verb forms in every sentence

```python
rows = []
for number, sentence in enumerate(sentences):
    past = 0
    present = 0
    for token in sentence:
        if "Past" in token.morph.get("Tense"):
            past = past + 1
        if "Pres" in token.morph.get("Tense"):
            present = present + 1
    rows.append({"sentence": number, "past": past, "present": present, "text": sentence.text})

tenses = pd.DataFrame(rows)
tenses.head(10)
```

```python
# the sentences with more past than present forms; pd.set_option() shows the whole text in each cell
pd.set_option("display.max_colwidth", None)
tenses[tenses["past"] > tenses["present"]]
```

These are mostly the sentences about the past of the Theater Bremen: Börgerding referring back to Hübner, who was dismissed after eleven years; Senator Thape, who pushed for the dismissal; the ensemble of more than 30 actors and its internal quarrels. A simple grammatical feature separates the history of the institution from the description of the evening

But not completely. "Hätte man ihm ... den Stuhl vor die Tür gesetzt" is a compound tense: only "hätte" carries the tense, and "würden" counts as past, although it is a subjunctive. The subjunctive (`Mood=Sub`) is a clue of its own: German uses it for reported speech, and at the end of the text the "Kritikenrundschau" reports what other critics wrote

```python
for token in doc:
    if "Sub" in token.morph.get("Mood"):
        print(token.i, token.text)
```

### 5.

The parser analyses the structure of each sentence. Every token gets a head, the word it depends on, and a label that names the relation. The labels of the German model come from the TIGER corpus: `sb` is the subject, `oa` the accusative object, `nk` a word that belongs to a noun, `mo` a modifier

```python
# the sentence about Senator Thape
sentence = sentences[6]
print(sentence.text)

for token in sentence:
    print(token.text, token.dep_, token.head.text)
```

```python
# displacy draws the structure as a tree of arcs
from spacy import displacy

displacy.render(sentence, style="dep", jupyter=True, options={"compact": True, "distance": 100})
```

With subject, verb and object we can ask: who does what? For every subject we look at its head (usually the verb) and at the accusative objects of that verb (`children` are the tokens that depend on a token)

```python
for token in doc:
    if token.dep_ == "sb":
        objects = []
        for child in token.head.children:
            if child.dep_ == "oa":
                objects.append(child.text)
        print(token.text, "|", token.head.lemma_, "|", objects)
```

Many subjects are pronouns ("der", "er", "sie"), and to know who they are we would have to go back in the text. But some lines are telling: "der – betreiben – Absetzung" is the Senator who brought about Hübner's dismissal; "Schauspieler – schicken – Grünewald" is wrong, though: in "Fünf Schauspieler schickte Grünewald auf die Suche" the director sends the actors. German word order is free, and the model guessed the wrong way round

The parser also helps with time. Expressions like "vor einem halben Jahrhundert" are a preposition and the words that depend on it; `subtree` gives us a token together with everything that depends on it

```python
time_words = ["Jahr", "Jahrhundert", "Spielzeit", "Epoche", "Zeit"]

for token in doc:
    if token.lemma_ in time_words and token.head.pos_ == "ADP":
        phrase = []
        for word in token.head.subtree:
            phrase.append(word.text)
        print(" ".join(phrase))
```

"vor einem halben Jahrhundert", "nach elf Jahren", "seit der laufenden Spielzeit", "in diesem Jahr": the time structure of the institution's history. Once again the parser is not perfect: it attaches "Inhaber dieses Postens" to "seit der laufenden Spielzeit"

### 6.

The last step: named entities. The German model knows four kinds: persons (`PER`), places (`LOC`), organisations (`ORG`) and everything else (`MISC`)

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# displacy marks the entities in the text
displacy.render(doc, style="ent", jupyter=True)
```

```python
rows = []
for ent in doc.ents:
    rows.append({"text": ent.text, "lemma": ent.lemma_, "label": ent.label_})

ents = pd.DataFrame(rows)
ents["label"].value_counts()
```

```python
# the persons; we count the lemmas, so that "Hübners" counts as "Hübner"
ents[ents["label"] == "PER"]["lemma"].value_counts()
```

The persons are recognised quite well. "Kurt Hübner" and "Hübner" are still counted separately, and "Schauspielensembles" is not a person. The other categories are worse: "Theater Bremen" is a place (it should be an organisation), and so are "Oral History", "Zeitzeugen" and "Ionesco". Only two organisations are found, the "Regionalliga" and the "Bundesliga", which here are metaphors. The model learnt its entities from Wikipedia, where an institution like a theatre is rare and a football league is common. For questions about institutions, this is the weakest step of the pipeline

### 7.

What can the pipeline tell us about institutional change? A few simple questions, each combining the steps from above

**Who holds which position?** In German, a title or role often stands directly before a name ("Kultursenator Moritz Thape"). We look at the token before every person

```python
# ent.start is the position of the first token of the entity; doc[ent.start - 1] is the token before it
for ent in doc.ents:
    if ent.label_ == "PER" and ent.start > 0:
        before = doc[ent.start - 1]
        if before.pos_ in ["NOUN", "PROPN"]:
            print(before.text, "|", ent.text)
```

This finds "Kultursenator Moritz Thape", "Senator Thape" and "Alt-68er Olaf Dinné", but also "Gespenst Kurt Hübner". It misses "Michael Börgerding, seit der laufenden Spielzeit Inhaber dieses Postens", because the role follows the name. Each pattern finds only what it describes

**Which words of the institution occur, and where?** We search the lemmas for a short list of words. With `in` we also find compounds like "Kultursenator" or "Schauspielensemble"

```python
institution_words = ["intendan", "spielzeit", "ensemble", "senator", "posten", "absetzung", "erbe", "epoche"]

for token in doc:
    for word in institution_words:
        if word in token.lemma_.lower():
            print(token.i, token.text, "|", token.sent.text[:120])
```

**When is the text talking about the past?** We already have this from step 4: the sentences in the past tense, together with the time expressions from step 5

Put together, a short history of the institution emerges from the text: an Intendant who made the theatre famous and was dismissed after eleven years by the Senator for Culture, an ensemble of more than thirty actors, and a new Intendant who, half a century later, measures himself against his predecessor. All of this is in the text, and for a single review we could have found it by reading. But each step has shown us its errors, and we could only notice them because we had read the text.
