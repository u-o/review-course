### 1. Dictionaries

A list finds a value by its position: `people[0]`.

Often we would rather look a value up by a name: what is the title, who is the critic?

A **dictionary** holds pairs of **keys** and **values**, like a dictionary of words in which we look up a word (the key) to find its meaning (the value).

It is written with curly brackets; each key is followed by a colon and its value, and the pairs are separated by commas

````python
review = {
    "critic": "Monty Jacobs",
    "newspaper": "Vossische Zeitung",
    "play": "Die Dreigroschenoper",
    "year": 1928,
}
print(review)
print(type(review))
````

````python
# square brackets with a key give us its value
print(review["critic"])
print(review["year"] + 1)
````

````python
# a key that does not exist: KeyError
print(review["director"])
````

````python
# in checks whether a key exists; len() counts the pairs; keys() and values() give all keys or all values
print("director" in review)
print(len(review))
print(review.keys())
print(review.values())
````

````python
# a new pair is added by giving a value to a new key; an existing key gets a new value the same way
review["director"] = "Erich Engel"
review["newspaper"] = "Vossische Zeitung, Berlin"
print(review)
````

````python
# values can also be lists
review["team"] = ["Erich Engel", "Caspar Neher", "Kurt Weill"]
print(review["team"])
print(review["team"][0])
````

````python
# a for-loop over a dictionary goes through its keys
for key in review:
    print(key, ":", review[key])
````

### 2. Counting with a dictionary

A dictionary is the best tool for counting many different things at once: every thing becomes a key, and its number the value.

Watch the dictionary grow in every round

````python
names = ["Brecht", "Weigel", "Brecht", "Engel", "Brecht", "Weigel"]

counts = {}
for name in names:
    if name in counts:
        counts[name] = counts[name] + 1     # we have seen the name before: add 1
    else:
        counts[name] = 1                    # the name is new: start with 1
    print(name, "→", counts)

print(counts)
````

### 3. Many records: a list of dictionaries

One dictionary describes one thing, e.g. one production. For many things we put many dictionaries into a list: a **list of dictionaries**.

Here are the first performances (*Uraufführungen*) of sixteen plays by Brecht. The director is given only for some of them.

Look closely: this is already a table in disguise. Every dictionary is a row, every key a column

````python
premieres = [
    {"title": "Trommeln in der Nacht", "year": 1922, "city": "München", "country": "Deutschland", "director": "Otto Falckenberg"},
    {"title": "Baal", "year": 1923, "city": "Leipzig", "country": "Deutschland"},
    {"title": "Im Dickicht der Städte", "year": 1923, "city": "München", "country": "Deutschland"},
    {"title": "Mann ist Mann", "year": 1926, "city": "Darmstadt", "country": "Deutschland"},
    {"title": "Die Dreigroschenoper", "year": 1928, "city": "Berlin", "country": "Deutschland", "director": "Erich Engel"},
    {"title": "Aufstieg und Fall der Stadt Mahagonny", "year": 1930, "city": "Leipzig", "country": "Deutschland"},
    {"title": "Die Rundköpfe und die Spitzköpfe", "year": 1936, "city": "Kopenhagen", "country": "Dänemark"},
    {"title": "Die Gewehre der Frau Carrar", "year": 1937, "city": "Paris", "country": "Frankreich"},
    {"title": "Mutter Courage und ihre Kinder", "year": 1941, "city": "Zürich", "country": "Schweiz", "director": "Leopold Lindtberg"},
    {"title": "Der gute Mensch von Sezuan", "year": 1943, "city": "Zürich", "country": "Schweiz", "director": "Leonard Steckel"},
    {"title": "Leben des Galilei", "year": 1943, "city": "Zürich", "country": "Schweiz", "director": "Leonard Steckel"},
    {"title": "Herr Puntila und sein Knecht Matti", "year": 1948, "city": "Zürich", "country": "Schweiz"},
    {"title": "Der kaukasische Kreidekreis", "year": 1948, "city": "Northfield", "country": "USA"},
    {"title": "Die Tage der Commune", "year": 1956, "city": "Karl-Marx-Stadt", "country": "DDR"},
    {"title": "Der aufhaltsame Aufstieg des Arturo Ui", "year": 1958, "city": "Stuttgart", "country": "BRD", "director": "Peter Palitzsch"},
    {"title": "Die heilige Johanna der Schlachthöfe", "year": 1959, "city": "Hamburg", "country": "BRD", "director": "Gustaf Gründgens"},
]

print(len(premieres))
print(premieres[0])
print(premieres[0]["title"])
````

````python
# a loop goes through the dictionaries one by one
for premiere in premieres:
    print(premiere["year"], premiere["title"])
````

````python
# with an if-clause we pick some of them
for premiere in premieres:
    if premiere["city"] == "Zürich":
        print(premiere["year"], premiere["title"])
````

````python
# where are Brecht's plays first performed? we count the countries with a dictionary
countries = {}
for premiere in premieres:
    country = premiere["country"]
    if country in countries:
        countries[country] = countries[country] + 1
    else:
        countries[country] = 1
print(countries)
````

````python
# a list of pairs is also common, e.g. the team of a production: each inner list holds a role and a name
team = [["Regie", "Erich Engel"], ["Bühne", "Caspar Neher"], ["Musik", "Kurt Weill"]]

# "role, name" unpacks each pair into two variables
for role, name in team:
    print(role, "→", name)
````

### 4. Libraries

Counting with loops works, but it takes many lines. For tables there are better tools.

Python itself is small, but many people have written code for all sorts of tasks and published it as **libraries**.

With `import` we make a library available, and call its functions with the name of the library and a dot

````python
# math is a library for mathematics; sqrt() is its square root function
import math
print(math.sqrt(16))
````

````python
# pandas is the library for tables; with "as" it gets a short name, and by convention everybody calls it pd
import pandas as pd
````

### 5. Tables with pandas

A table in pandas is called a **DataFrame**.

From a list of dictionaries it is made in one line

````python
df = pd.DataFrame(premieres)
df
````

Every dictionary has become a **row**, every key a **column**.

On the left pandas numbers the rows, starting at 0 (the **index**).

Where a dictionary had no `director`, the table shows `NaN` (*not a number*): the cell is empty

````python
# shape gives the number of rows and columns; columns gives the names of the columns
print(df.shape)
print(df.columns)
````

````python
# head() shows only the first rows (5, or as many as we say)
df.head(3)
````

````python
# a single column is picked by its name, like a value in a dictionary
df["city"]
````

````python
# value_counts() counts the values of a column: what took us a loop and an if-clause above is now one line
df["country"].value_counts()
````

````python
# several columns: a LIST of names inside the brackets, hence the double brackets
df[["year", "title"]]
````

Picking rows is the step that looks most strange at first. We go slowly.

First a comparison with a whole column: pandas compares every value and gives back a column of `True` and `False`

````python
df["country"] == "Schweiz"
````

````python
# put into the brackets of df, this column keeps only the rows with True
swiss = df[df["country"] == "Schweiz"]
swiss
````

````python
# the same with other comparisons
df[df["year"] < 1933]
````

````python
# isin() keeps the rows whose value is in a list
df[df["city"].isin(["Berlin", "München"])]
````

````python
# isna() finds the empty cells; sum() counts them per column (True counts as 1)
df.isna().sum()
````

````python
# notna() is the opposite: here the rows that have a director
df[df["director"].notna()]
````

````python
# a new column is calculated from an existing one
# // divides and drops the rest: 1928 // 10 = 192, times 10 = 1920
df["decade"] = df["year"] // 10 * 10
df[["year", "decade", "title"]].head()
````

````python
# a new column can also come from a loop: one value per row, collected in a list, in the same order as the rows
periods = []
for year in df["year"]:
    if year < 1933:
        periods.append("before the exile")
    elif year < 1949:
        periods.append("during the exile")
    else:
        periods.append("after the return")

df["period"] = periods
df[["year", "period", "city"]]
````

````python
df["period"].value_counts()
````

````python
# sort_values() sorts the rows; ascending=False sorts from large to small
df.sort_values("year", ascending=False).head()
````

````python
# plot() draws a chart; sort_index() puts the decades in order
import matplotlib.pyplot as plot
df["decade"].value_counts().sort_index().plot(kind="bar", title="Brecht premieres per decade")
````

The table shows in a few lines what was a whole chapter of Brecht's career: until 1933 his plays are first performed in German cities, during the exile in Zürich, Copenhagen, Paris and the USA, and after the war in the two German states.

Questions about institutions (where, by whom, when) become questions we can ask a table.

### 6. Back to the text: a table of words

The same steps (an empty list, a loop, one dictionary per item, a DataFrame) turn a text into a table.

This is exactly how we will work with the results of language processing later: one row per word

````python
text = "Es gibt in dieser Chronik keine Menschen, die restlos gut, und keine, die restlos böse sind. Menschen sind sie, ausgeliefert dem Kriege."

rows = []
for position, word in enumerate(text.split()):
    clean = word.strip(".,")
    rows.append({
        "position": position,
        "word": clean,
        "length": len(clean),
        "capital": clean[0].isupper(),
    })

words = pd.DataFrame(rows)
words.head(10)
````

````python
# the most frequent words
words["word"].value_counts().head()
````

````python
# only the words with a capital letter
words[words["capital"] == True]
````

### Summary

| concept | example |
|---|---|
| a dictionary | `review = {"critic": "Monty Jacobs", "year": 1928}`, `review["critic"]` |
| add or change a pair | `review["director"] = "Erich Engel"` |
| keys, in, loop | `review.keys()`, `"director" in review`, `for key in review:` |
| counting with a dictionary | `if name in counts: … else: …` |
| a list of dictionaries | `[{"title": …}, {"title": …}]`: a table in disguise |
| unpacking pairs | `for role, name in team:` |
| a library | `import pandas as pd` |
| a DataFrame | `df = pd.DataFrame(premieres)`, `df.shape`, `df.head()` |
| columns | `df["city"]`, `df[["year", "title"]]`, `df["decade"] = df["year"] // 10 * 10` |

| picking rows | `df[df["year"] < 1933]`, `df[df["city"].isin([…])]`, `df[df["director"].notna()]` |
| sorting, plotting | `df.sort_values("year")`, `….plot(kind="bar")` |
